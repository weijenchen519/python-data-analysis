# 項目：評估和清理英國電商公司銷售數據

## 分析目標

此數據分析的目的是，根據市場銷售數據，挖掘暢銷產品，以便制定更有效的市場策略來提升營收。

本實戰項目的目的在於練習評估數據乾淨和整潔度，並且基於評估結果，對數據進行清洗，從而得到可供下一步分析的數據。

## 簡介

原始資料集記錄了一家英國線上零售公司在2010年12月1日至2011年12月9日期間的所有交易情況，涵蓋了該公司在全球不同國家和地區的業務數據。該公司主要銷售涵蓋各個場景的禮品，包括但不限於生日禮品、結婚紀念品、聖誕禮品等等。該公司的客戶群體主要包括批發商和個人消費者，其中批發商佔據了相當大的比例。

資料每列的含義如下：

InvoiceNo：發票號碼。6位數，作為交易的唯一識別符。如果這個代碼以字母「C」開頭，表示這筆交易被取消。

StockCode：產品代碼。5位數，作為產品的唯一識別符。

Description：產品名稱。

Quantity：產品在交易中的數量。

InvoiceDate：發票日期和時間。交易發生的日期和時間。

UnitPrice：單價。價格單位為英鎊（£）。

CustomerID：客戶編號。5位數，作為客戶的唯一識別符。

Country：國家名稱。客戶所居住的國家的名稱。

## 讀取數據

```python
import pandas as pd
```

```python
original_data = pd.read_csv("e_commerce.csv")
```

```python
original_data.head()
```

## 評估數據

```python
original_data.sample(15)
```

從抽樣的15行數據數據來看，數據符合「每列是一個變量，每行是一個觀察值，每個單元格是一個值」，具體來看每行是關於某商品的一次交易，每列是交易相關的各個變量，因此不存在結構性問題。

<font size="5">評估數據乾淨度</font>

```python
original_data.info()
```

從輸出結果來看，數據共有541909條觀察值，而Description、CustomerID變量存在缺失值。

此外，InvoiceDate的數據類型應為日期，CustomerID的數據類型應為字串，應當進行數據格式轉換。

<font size = "5">評估缺失數據</font>

```python
original_data[original_data['Description'].isnull()]
```

查看是否有Description有缺失但Unit price不為0的數據

```python
original_data[(original_data['Description'].isnull()) & (original_data['UnitPrice'] != 0) ]
```

Description有缺失但Unit price都為0,且對於我們分析暢銷產品並無用處，後續可刪除

CustomerID變量同样存在缺失值，因此也根據條件提取出缺失觀察值。

```python
original_data[(original_data['Description'].isnull()) & (original_data['CustomerID'].isnull()) ]
```

CustomerID的缺失直不影響我們對於暢銷產品的分析，故保留

<font size = "5">評估重複數據</font>

```python
根據數據變量的含義來看，雖然InvoiceNo、StockCode和CustomerID都是唯一識別符，但一次交易可能包含多件商品，因此InvoiceNo可以存在重複；
不同交易可以包含同件商品，因此StockCode可以存在重複；
顧客可以進行多次交易或下單多個商品，因此CustomerID也可以存在重複。
那麼針對此數據集，我們無需評估重複數據。
```

<font size = "5">評估不一致數據</font>

```python
不一致數據可能存在於Country變量中，我们要查看是否存在多个不同值指代同一國家的情况。
```

```python
original_data['Country'].value_counts()
```

從Country變量值來看，"USA"、"United States"均表示美國，"United Kingdom"、"UK"、"U.K."均表示英國，
因此應該在清洗步驟中對這些值進行統一，只保留一個指代值。

<font size = "5">評估錯誤數據</font>

```python
original_data.describe()
```

從輸出結果來看，Quantity和UnitPrice存在負數，會對後續數值分析產生影響。

因此，我們先篩選出Quantity數值為負數的觀察值，進一步評估其含義。

```python
original_data[original_data['Quantity']<0]
```

```python
original_data[(original_data['Quantity']<0) & (original_data['InvoiceNo'].str[0]!='C') ]
```

以上猜想被證實錯誤，因為還存在Quantity變量為負數且InvoiceNo不以C開頭的觀察值。

但根據以上輸出結果，這些篩選出的觀察值的UnitPrice觀察值均為0，因此增加UnitPrice的條件進行驗證。

```python
original_data[(original_data['Quantity']<0) & (original_data['InvoiceNo'].str[0]!='C') & (original_data['UnitPrice'] != 0) ]
```

根據輸出結果，猜想得到驗證，當Quantity變量為負數時，觀察值滿足以下條件之一：

InvoiceNo以C開頭，表示訂單被取消

UnitPrice為0，說明單價為0英鎊

這些交易數據均不是有效成交數據，不貢獻銷售，不在後續分析範圍內，因此我們將在數據清理步驟中，將Quantity為負數的觀察值刪除。

接下來篩選出UnitPrice為負數的觀察值。

```python
original_data[original_data['UnitPrice'] < 0 ]
```

```python
從輸出結果來看，UnitPrice為負數的觀察值都是壞帳調帳，不屬於實際商品交易數據，因此在數據清理步驟中也將其刪除。
```

## 清理數據

根據前面評估部分得到的結論，我們需要進行的數據清理包括：

把InvoiceDate變量的數據類型轉換為日期時間

把CustomerID變量的數據類型轉換為字串

把Description變量缺失的觀察值刪除

把Country變量值"USA"替換為"United States"

把Country變量值"UK"、"U.K."替換為"United Kingdom"

把Quantity變量值為負數的觀察值刪除

把UnitPrice變量值為負數的觀察值刪除

為了區分開經過清理的數據和原始的數據，我們創建新的變量cleaned_data，讓它為original_data複製出的副本。我們之後的清理步驟都將被運用在cleaned_data上。

```python
cleaned_data = original_data.copy()
cleaned_data.head()
```

把InvoiceDate變量的數據類型轉換為日期時間類型：

```python
cleaned_data['InvoiceDate'] = pd.to_datetime(cleaned_data['InvoiceDate'])
cleaned_data['InvoiceDate']
```

把CustomerID變量的數據類型轉換為字符串

```python
cleaned_data['CustomerID'] = cleaned_data['CustomerID'].astype(str)
cleaned_data['CustomerID'] 
```

把CustomerID變量值結尾的".0"删除：

```python
cleaned_data['CustomerID'].str.slice(0,-2)
```

把Description變量缺失的觀察值刪除，並查看刪除後該列空缺值個數和：

```python
cleaned_data.dropna(subset=['Description'], inplace= True )
```

```python
cleaned_data['Description'].isnull().sum()
```

把Country变量值"USA"替换为"United States"

```python
cleaned_data['Country'] = cleaned_data['Country'].replace({'USA':'United States'})
```

```python
len(cleaned_data[cleaned_data['Country']== 'USA'])
```

把Country变量值"UK"、"U.K."替换为"United Kingdom"

```python
cleaned_data['Country'] = cleaned_data['Country'].replace({'UK':'United Kingdom','U.K.':'United Kingdom'})
```

```python
print(len(cleaned_data[cleaned_data['Country']== 'U.K.']))
print(len(cleaned_data[cleaned_data['Country']== 'UK']))
```

把Quantity变量值為負數的觀察值删除(這邊反過來想，把Quantity>=0的值保留，即把<0的值刪除)

```python
cleaned_data = cleaned_data[cleaned_data['Quantity'] >= 0 ]
```

```python
len(cleaned_data[cleaned_data['Quantity'] < 0 ])
```

把UnitPrice变量值為負數的觀察值删除，檢查替換後UnitPrice變量為0的個數:

```python
cleaned_data = cleaned_data[cleaned_data['UnitPrice'] >= 0 ]
```

```python
len(cleaned_data[cleaned_data['UnitPrice'] < 0 ])
```

## 保存清理後的數據

完成數據清理後，把乾淨整齊的數據保存到新的文件里，文件名為e_commerce_cleaned.csv。

```python
cleaned_data.to_csv('e_commerce_cleaned.csv', index = False)
```
